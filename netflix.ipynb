```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv("netflix_titles.csv")
```

```python
df=df.fillna("Unknown") 
```

```python
df.drop_duplicates()
```

```python
df.head()
```

```python
df.groupby("country").size()
```

```python
df["country"] = df["country"].str.split(",")
df = df.explode("country")
df["country"] = df["country"].str.strip()
df["country"] = df["country"].replace({"": "Unknown"})
countries = df.groupby("country").size().sort_values(ascending=False)
```

```python
top_10 = countries.head(10)
fig, ax = plt.subplots(figsize=(10, 6))
bars = ax.bar(top_10.index, top_10.values, color='steelblue')
ax.bar_label(bars, padding=3)
ax.set_title('Top 10 Countries')
ax.set_xlabel('Country')
ax.set_ylabel('Count')
plt.xticks(rotation=45, ha='right')
```

```python
movieVSshow = df.groupby("type").size()
```

```python
labels = ['movies', 'tv shows']
colors = ['#4C72B0', '#DD8452']
fig, ax = plt.subplots(figsize=(6, 6))
ax.pie(
    movieVSshow,
    labels=labels,
    colors=colors,
    autopct='%1.1f%%',
    startangle=90
)
ax.set_title('movie VS show')
plt.tight_layout()
plt.show()
```

```python
year = df.groupby("release_year").size()
```

```python
peak_year_movie = year.idxmax()
peak_count_movie = year.max()
fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(year.index, year.values, marker='o', linestyle='-', color='steelblue', linewidth=2)
ax.scatter([peak_year_movie], [peak_count_movie], color='crimson', s=100, zorder=5)
ax.annotate(
    f'Peak: {peak_year_movie}\n({peak_count_movie} releases)',
    xy=(peak_year_movie, peak_count_movie),
    xytext=(peak_year_movie - 30, peak_count_movie-60),  # Positions label slightly to the left
    arrowprops=dict(facecolor='black', shrink=0.08, width=1.5, headwidth=5),
    fontsize=10,
    fontweight='bold',
    color='crimson'
)
ax.set_title('Releases Over Time')
ax.set_xlabel('Release Year')
ax.set_ylabel('Count')
ax.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
duration = df[df['type'] == 'Movie'].groupby("duration").size()
```

```python
clean_duration = duration.drop('Unknown', errors='ignore')
minutes = clean_duration.index.str.replace(' min', '', regex=False).astype(int)
counts = clean_duration.values
bins = range(0, 340, 30)
fig, ax = plt.subplots(figsize=(10, 6))
n, bins, patches = ax.hist(
    minutes,
    bins=bins,
    weights=counts,
    color='steelblue',
    edgecolor='black'
)
ax.bar_label(patches, padding=3)
ax.set_title('Distribution of Durations')
ax.set_xlabel('Duration (minutes)')
ax.set_ylabel('Count')
ax.set_xticks(bins)
plt.tight_layout()
plt.show()
```

```python
seasons = df[df['type'] == 'TV Show'].groupby("duration").size()
```

```python
clean_seasons = seasons.drop('Unknown', errors='ignore')
season_nums = clean_seasons.index.str.extract(r'(\d+)')[0].astype(int)
season_series = pd.Series(clean_seasons.values, index=season_nums).sort_index()
fig, ax = plt.subplots(figsize=(10, 6))
ax.plot(
    season_series.index,
    season_series.values,
    marker='o',
    linestyle='-',
    color='steelblue',
    linewidth=2
)
for x_val, y_val in zip(season_series.index, season_series.values):
    ax.annotate(
        str(y_val),
        xy=(x_val, y_val),
        xytext=(0, 6),
        textcoords='offset points',
        ha='center',
        fontsize=9
    )
ax.set_title('TV Shows by Number of Seasons')
ax.set_xlabel('Number of Seasons')
ax.set_ylabel('Count')
ax.set_xticks(season_series.index)  # Shows every season number clearly on the X-axis
ax.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
dates = pd.to_datetime(df['date_added'].str.strip(), errors='coerce')
temp_df = pd.DataFrame({
    'year': dates.dt.year,
    'month': dates.dt.month
}).dropna().astype(int)
heatmap_data = temp_df.groupby(['year', 'month']).size().unstack(fill_value=0)
heatmap_data = heatmap_data.reindex(columns=range(1, 13), fill_value=0)  # Ensures all 12 months appear
month_names = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
               'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
fig, ax = plt.subplots(figsize=(12, 6))
im = ax.imshow(heatmap_data.values, cmap='YlGnBu', aspect='auto')
cbar = fig.colorbar(im, ax=ax)
cbar.set_label('Count Added')
ax.set_xticks(np.arange(12))
ax.set_xticklabels(month_names)
ax.set_yticks(np.arange(len(heatmap_data.index)))
ax.set_yticklabels(heatmap_data.index)
for i in range(len(heatmap_data.index)):
    for j in range(12):
        val = heatmap_data.values[i, j]
        text_color = 'white' if val > heatmap_data.values.max() / 2 else 'black'
        ax.text(j, i, str(val), ha='center', va='center', color=text_color, fontsize=9)
ax.set_title('Movies & TV Shows Added by Month and Year')
ax.set_xlabel('Month')
ax.set_ylabel('Year')
plt.tight_layout()
plt.show()
```
